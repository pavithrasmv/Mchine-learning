```python
import random

states = ["A", "B", "C", "D", "Goal"]

actions = ["Left", "Right"]

rewards = {
    "A": {"Left": 0, "Right": 0},
    "B": {"Left": 0, "Right": 0},
    "C": {"Left": 0, "Right": 0},
    "D": {"Left": 0, "Right": 10}
}

Q = {s: {a: 0 for a in actions} for s in states}

alpha = 0.5
gamma = 0.9
epsilon = 0.3

for episode in range(30):

    state = "A"

    while state != "Goal":

        # Choose action: explore or exploit
        if random.random() < epsilon:
            action = random.choice(actions)
        else:
            action = max(Q[state], key=Q[state].get)

        # Find next state
        if state == "A" and action == "Right":
            next_state = "B"

        elif state == "B" and action == "Right":
            next_state = "C"

        elif state == "C" and action == "Right":
            next_state = "D"

        elif state == "D" and action == "Right":
            next_state = "Goal"

        else:
            next_state = state   # Left keeps the agent in the same state

        # Get reward
        reward = rewards.get(state, {}).get(action, 0)

        # Current Q-value
        old_value = Q[state][action]

        # Maximum Q-value of next state
        next_max = max(Q[next_state].values()) if next_state != "Goal" else 0

        # Q-learning update
        Q[state][action] = old_value + alpha * (
            reward + gamma * next_max - old_value
        )

        # Move to next state
        state = next_state


print("Learned Q-values:")

for s in Q:
    print(s, Q[s])
```
