```python
import nltk

# Download required datasets
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# ---------------------------------------
# Step 1: Input text
# ---------------------------------------
text = "I love teaching NLP. NLP is very interesting and I love my students."

print("Original text:")
print(text)

# ---------------------------------------
# Step 2: Tokenization
# ---------------------------------------
tokens = word_tokenize(text)

print("\nWord tokens:")
print(tokens)

# ---------------------------------------
# Step 3: Remove stopwords
# ---------------------------------------
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words and word.isalpha()
]

print("\nTokens after removing stopwords:")
print(filtered_tokens)

# ---------------------------------------
# Step 4: Word Frequency
# ---------------------------------------
freq_dist = FreqDist(filtered_tokens)

print("\nWord frequencies:")
for word, freq in freq_dist.items():
    print(word, ":", freq)

# ---------------------------------------
# Step 5: Most Common Words
# ---------------------------------------
print("\nMost common words:")
print(freq_dist.most_common(3))

```

```python
import nltk

# Download required data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.corpus import stopwords

# Input text
text = "Python is a programming language and it is easy to learn."

# Convert text into words
words = nltk.word_tokenize(text)

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords
filtered_words = []

for word in words:
    if word.lower() not in stop_words:
        filtered_words.append(word)

# Display results
print("Original Text:")
print(text)

print("\nWords:")
print(words)

print("\nAfter Stopword Removal:")
print(filtered_words)
```

```python
import nltk

# Download required data
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.stem import PorterStemmer

# Create stemmer
stemmer = PorterStemmer()

# Input text
text = "The students are studying and playing games. They studied yesterday."

# Tokenization
words = nltk.word_tokenize(text)

# Stemming
stemmed_words = []

for word in words:
    stemmed_words.append(stemmer.stem(word))

# Display result
print("Original Words:")
print(words)

print("\nAfter Stemming:")
print(stemmed_words)
```

```python
import nltk

# Download required packages
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

# Words
words = ["playing", "played", "studies", "children", "cars"]

# Lemmatization
for word in words:
    result = lemmatizer.lemmatize(word)
    print(word, "→", result)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.classify import NaiveBayesClassifier

# Training data
data = [
    ("Win a free lottery prize", "spam"),
    ("Congratulations you won money", "spam"),
    ("Claim your free gift", "spam"),
    ("Get free cash now", "spam"),

    ("Can we meet tomorrow", "not_spam"),
    ("Please send me the report", "not_spam"),
    ("Let's have lunch today", "not_spam"),
    ("The meeting is at 10 AM", "not_spam")
]

# Convert sentence into words
def extract_features(sentence):
    words = nltk.word_tokenize(sentence.lower())
    return {word: True for word in words}

# Create training features
training_data = []

for sentence, label in data:
    features = extract_features(sentence)
    training_data.append((features, label))

# Train the model
classifier = NaiveBayesClassifier.train(training_data)

# Test sentences
test_sentences = [
    "free money prize",
    "please attend the meeting",
    "claim your free cash",
    "send the report tomorrow"
]

# Prediction
for sentence in test_sentences:
    features = extract_features(sentence)
    result = classifier.classify(features)

    print(sentence, "→", result)
```

```python
import nltk

# Download VADER
nltk.download('vader_lexicon')

from nltk.sentiment import SentimentIntensityAnalyzer

# Create analyzer
sia = SentimentIntensityAnalyzer()

# Test sentences
sentences = [
    "I love this product.",
    "This product is very good.",
    "This product is terrible.",
    "I hate this service.",
    "The product is okay."
]

# Analyze sentiment
for sentence in sentences:

    score = sia.polarity_scores(sentence)

    if score['compound'] >= 0.05:
        sentiment = "Positive"
    elif score['compound'] <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(sentence)
    print("Score:", score)
    print("Sentiment:", sentiment)
    print()
```
