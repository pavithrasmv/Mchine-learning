```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
# Step 1: Create dataset
X = []
y = []
for i in range(1, 51):
  X.append([i, i+1, i+2])   # input sequence
  y.append(i+3)
X = np.array(X)
y = np.array(y)
# next value
# Step 2: Normalize data (VERY IMPORTANT)--- convert them into smaller units
X = X / 100.0
y = y / 100.0
# Step 3: Reshape for RNN [samples, timesteps, features]--- sees numbers one by one
X = X.reshape((X.shape[0], X.shape[1], 1))
# Step 4: Build model
model = Sequential()
model.add(SimpleRNN(20, activation='relu', input_shape=(3,1)))
model.add(Dense(1))
# Step 5: Compile
model.compile(optimizer='adam', loss='mse')
# Step 6: Train
model.fit(X, y, epochs=300, verbose=0)
print("Training completed.")
# Step 7: Test
test_input = np.array([[51, 52, 53]]) / 100.0
test_input = test_input.reshape((1, 3, 1))
prediction = model.predict(test_input)
# Step 8: Denormalize output
prediction = prediction * 100
print("Predicted next number:", prediction[0][0])
```

```python

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense
# Step 1: Create dataset
X = []
y = []
for i in range(1, 51):
  X.append([i, i+1, i+2])
  y.append(i+3)
X = np.array(X)
y = np.array(y)
# Step 2: Normalize
X = X / 100.0
y = y / 100.0
# Step 3: Reshape
X = X.reshape((X.shape[0], X.shape[1], 1))
# Step 4: Build GRU model
model = Sequential()
model.add(GRU(20, input_shape=(3,1)))
model.add(Dense(1))
# Step 5: Compile
model.compile(optimizer='adam', loss='mse')
# Step 6: Train
model.fit(X, y, epochs=300, verbose=0)
print("Training completed.")
# Step 7: Test
test_input = np.array([[51, 52, 53]]) / 100.0
test_input = test_input.reshape((1, 3, 1))
prediction = model.predict(test_input)
# Step 8: Denormalize
prediction = prediction * 100
print("Predicted next number:", prediction[0][0])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, MultiHeadAttention, LayerNormalization, GlobalAveragePooling1D
# Step 1: Create dataset
X = []
y = []
for i in range(1, 51):
  X.append([i, i+1, i+2])
  y.append(i+3)
X = np.array(X)
y = np.array(y)
# Step 2: Normalize
X = X / 100.0
y = y / 100.0
# Step 3: Reshape (samples, timesteps, features)
X = X.reshape((X.shape[0], X.shape[1], 1))
# Step 4: Transformer Model
inputs = Input(shape=(3,1))
# Attention Layer
attention = MultiHeadAttention(num_heads=2, key_dim=1)(inputs, inputs)
# Add & Normalize
x = LayerNormalization()(attention + inputs)
# Pooling
x = GlobalAveragePooling1D()(x)
# Output
outputs = Dense(1)(x)
model = Model(inputs, outputs)
# Step 5: Compile
model.compile(optimizer='adam', loss='mse')
model = Model(inputs, outputs)
# Step 5: Compile
model.compile(optimizer='adam', loss='mse')
# Step 6: Train
model.fit(X, y, epochs=300, verbose=0)
print("Training completed.")
# Step 7: Test
test_input = np.array([[51, 52, 53]]) / 100.0
test_input = test_input.reshape((1, 3, 1))
prediction = model.predict(test_input)
prediction = prediction * 100
print("Predicted next number:", prediction[0][0])
```
